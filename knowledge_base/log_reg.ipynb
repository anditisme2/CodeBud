```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Brain_Tumor.csv")
df
```

```python
x=np.array(df['Tumor_Size']).reshape(-1,1)
y=np.array(df['Class']).reshape(-1,1)
```

```python
#SPLIT THE DATASET USING train_test_split() method
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2)
```

```python
#IMPORT LOGISTIC REGRESSION MODEL
from sklearn.linear_model import LogisticRegression
```

```python
model_logreg=LogisticRegression()
```

```python
#FIT THE MODEL/ TRAIN THE MODEL
model_logreg.fit(x_train, y_train)
```

```python
#prediction of the model
y_pred=model_logreg.predict(x_test)
y_pred
```

```python
#IMPORT CLASSIFICATION METRICS
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
#accuracy
acc=accuracy_score(y_pred, y_test)
print(f"Accuracy of the proposed model: {acc}")
```

```python
co=confusion_matrix(y_pred, y_test)
```

```python
print(co)
```

```python
outputs=["malignant", "benign"]
cr=classification_report(y_pred, y_test, target_names=outputs)
print(cr)
```

```python
import seaborn as sns
```

```python
sns.heatmap(co, cmap="coolwarm", annot=True, xticklabels=[0,1], yticklabels=[0,1])
plt.xlabel("Actual Output")
plt.ylabel("Predicted Output")
plt.title("Brain Tumor")
plt.show()
```
