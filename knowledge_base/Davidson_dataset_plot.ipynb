```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("labeled_data.csv")
```

```python
counth=len(df[df['class']==0])
print(f"No. of hate speech tweets:{counth}")
```

```python
counto=len(df[df['class']==1])
print(f"No. of offensive tweets:{counto}")
```

```python
countn=len(df[df['class']==2])
print(f"No. of neutral tweets:{countn}")
```

```python
count=len(df['class'])
print(count)
```

```python
labels=['hate speech', 'offensive language', 'neutral']
count=[counth, counto, countn]
plt.title("Davidson dataset")
plt.bar(labels, count, color="skyblue", edgecolor="black", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
labels=['hate speech and offensive language', 'neutral']
counts=counth+counto
count=[counts, countn]
plt.bar(labels, count, color="skyblue", edgecolor="black", width=0.3, alpha=0.7)
plt.tight_layout
plt.show()
```
