```python
import numpy as np
```

```python
l1=df.values
l1
len(l1)
```

```python
#Euclidean Distance
def euclidean_distance_function(x, y):
    return np.sqrt(np.sum(x-y)**2)
```

```python
#Manhattan Distance
def manhattan_distance_function(a, b):
    return np.sum(np.abs(a - b))
```

```python
#Cosine Distance
def cosine_distance_function(a, b):
    return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
```

```python
#Minkowski Distance
def minkowski_distance_function(x, y, p):
    n=1/p
    return (np.sum(x-y)**p)**n
```

```python
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        print("euclidean distance", euclidean_distance_function(l1[i],l1[j]))
```

```python
#Manhattan DIstance
```

```python
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        print("manhattan distance", manhattan_distance_function(l1[i],l1[j]))
```

```python
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        print("cosine distance", cosine_distance_function(l1[i],l1[j]))
```

```python
#Mincowski Distance
```

```python
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        print("cosine distance", minkowski_distance_function(l1[i],l1[j], 2))
```
