```python
#Linear Regression Model  y=mx+c
def linear_regresssion_model(x, w, b):
    f_x=np.zeros(len(x))
    for i in range(len(x)):
        f_x[i]=w*x[i]+b
    return f_x
```

```python
linear_regression=linear_regresssion_model(x, w, b)
linear_regression
```

GRADIENT DESCENT

```python
def gradient_descent_cost_function(x, y, w, b):
    j_dw=0
    j_db=0
    for i in range(len(x)):
        f_x=w*x[i]+b
        j_dw=j_dw+(f_x-y[i])*x[i]
        j_db=j_db+(f_x-y[i])
    j_dw=j_dw/2
    j_db=j_db/2
    return j_dw, j_db
```
