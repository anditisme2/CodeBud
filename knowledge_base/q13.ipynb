```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("brain_tumor.csv")
```

```python
mean_val = df['Tumor_Size'].mean()
median_val = df['Tumor_Size'].median()
```

```python
sns.scatterplot(x=df.index, y='Tumor_Size', hue='Class', data=df)
plt.axhline(mean_val, color='red', linestyle='--', label='Mean')
plt.axhline(median_val, color='green', linestyle='-', label='Median')
plt.title('Tumor Size Analysis')
plt.legend()
plt.show()
```
