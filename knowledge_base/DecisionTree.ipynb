```python
import pandas as pd
```

```python
from sklearn import preprocessing
```

```python
df=pd.read_csv("play_tennis.csv")
```

```python
process_tennis=preprocessing.LabelEncoder()
```

```python
df=df.apply(process_tennis.fit_transform)
```

```python
from sklearn.model_selection import train_test_split
```

```python
df.drop("day", axis=1)
```

```python
features=["outlook", "temp", "humidity", "wind"]
```

```python
X=df[features]
Y=df["play"]
```

```python
x_train, x_test, y_train, y_test=train_test_split(X, Y, train_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model=DecisionTreeClassifier()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
acc=accuracy_score(y_pred, y_test)
acc
```

```python
co=confusion_matrix(y_pred, y_test)
co
```

```python
outputs=["not play", "play"]
cr=classification_report(y_pred, y_test, target_names=outputs)
print(cr)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.heatmap(co, cmap='coolwarm', annot=True, xticklabels=[0,1], yticklabels=[0,1])
plt.xlabel("Yes")
plt.ylabel("No")
plt.title("Tennis")
plt.show()
```
