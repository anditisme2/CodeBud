CPU time = how long it takes to execute in CPU  
Wall time = how long does it take in real time (CPU time + waiting time)

```python
torch_rand1 = torch.rand(100, 100, 100, 100).to(device)
torch_rand2 = torch.rand(100, 100, 100, 100).to(device)
np_rand1 = torch.rand(100, 100, 100, 100)
np_rand2 = torch.rand(100, 100, 100, 100)

#GPU
start_time = time.time()

rand = (torch_rand1 @ torch_rand2)

end_time = time.time()

elapsed_time = end_time - start_time
print(f"{elapsed_time: .8f}")

#CPU
start_time = time.time()

rand = np.multiply(np_rand1, np_rand2)

end_time = time.time()

elapsed_time = end_time - start_time
print(f"{elapsed_time: .8f}")
```

**nn.Embedding** is a class in PyTorch library that represents an embedding layer, used in NLP and other ML tasks to represent discrete inputs as continuous-valued vectors.

```python
import torch.nn as nn
# Initialize an embedding layer
vocab_size = 10000
embedding_dim = 100
embedding = nn.Embedding(vocab_size, embedding_dim)

# Create some input indices
input_indices = torch.LongTensor([1, 5, 3, 2])

# Apply the embedding layer
embedded_op = embedding(input_indices)

# o/p = tensor of shape (4, 100), where 4 = no. of inputs and 100 = dimensionality of the embedding vectors
print(embedded_op.shape)
```

**Actication Functions**:  
**Relu** *Rectified Linear Unit*  
ReLU(x) = max(0, x)  
**Sigmoid**  
Sigmoid(x) = 1/(1 + exp(-x))  
**Tanh**  
tanh(x) = (exp(x) - exp(-x))/(exp(x) + exp(-x))

```python
x = torch.tensor([0.5], dtype = torch.float32)
# y = F.sigmoid(x)
y = F.tanh(x)
print(y)
```
