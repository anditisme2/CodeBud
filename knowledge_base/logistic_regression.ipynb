```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Brain_Tumor.csv")
df
```

```python
x=np.array(df['Tumor_Size'])
y=np.array(df['Class'])
```

```python
plt.scatter(x,y, marker="x", c="r", label="Tumor")
plt.xlabel("Tumor Size")
plt.ylabel("Class")
plt.show()
```

```python
w=0
b=0
```

```python
def sigmoid(z):
    y=1/(1+np.exp(-z))
    return y
```

```python
#logistic regression
def logistic_regression(w, b, x):
    z=w*x+b
    f_p=sigmoid(z)
    return f_p 
```

```python
y_pred=logistic_regression(w, b, x)
print(y_pred)
```

```python
#cost function
def cost_function(x,y,w,b):
    fn=0
    m=len(x)
    for i in range (m):
        z=w*x[i]+b
        y_pred=sigmoid(z)
        epsilon=1*np.exp(-8)
        fn+=(-y[i]*np.log(y_pred+epsilon)-(1-y[i])*np.log(1-y_pred+epsilon))
        cf=fn/m
    return cf
```

```python
ans=cost_function(x,y,w,b)
print(ans)
```

```python
def gradient_descent(x, y, w, b):
    j_w=0
    j_b=0
    m=len(x)
    for i in range (m):
        z=w*x[i]+b
        y_pred=sigmoid(z)
        j_w+=(y_pred-y[i])*x[i]
        j_b+=(y_pred-y[i])
    j_w=j_w/m
    j_b=j_b/m
    return j_w, j_b
```

```python
x=(x-np.mean(x))/np.std(x)
w=0
b=0
iterations=1000
alpha=0.01
for i in range(iterations):
    j_dw, j_db= gradient_descent(x, y, w, b)
    w=w-alpha*j_dw
    b=b-alpha*j_db
print(f"j_dw={j_dw}, j_db={j_db}")
```

```python
x_range=np.linspace(min(x), max(x), 200)
y_curve=sigmoid(w * x_range + b)

#Scatter Plot
plt.scatter(x, y)
plt.plot(x_range, y_curve)
plt.xlabel("inputs")
plt.ylabel("outputs")
plt.show()
```
